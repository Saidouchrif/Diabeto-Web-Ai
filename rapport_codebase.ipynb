````python
# Rapport technique du projet Diabeto-Web

Ce notebook présente l’architecture, les principales fonctions et l’organisation du code du projet Diabeto-Web (Laravel + IA Python).
````

````python
## 1. Présentation du projet
- Plateforme de gestion de patients diabétiques
- Diagnostic automatique via IA (FastApi/Python)
- Interface web moderne (Laravel, Blade, Tailwind CSS)

````

````python
## 2. Architecture générale
```mermaid
graph TD;
  Utilisateur-->|Navigateur|Frontend(Blade/Tailwind)
  Frontend-->|HTTP|Laravel_Backend
  Laravel_Backend-->|API|IA_Python(FestApi)
  Laravel_Backend-->|SQL|DB[(MySQL)]
```

````

````python
## 3. Contrôleurs et routes Laravel
- **PatientController** : CRUD patients, prédiction IA, filtres
- **MedcianeController** : Dashboard
- **LoginController** : Authentification

### Exemples de routes
- `/home/medciane/patients` : liste
- `/home/medciane/details/{id}` : détails
- `/home/medciane/predict/{id}` : prédiction IA

````

````python
## 4. Modèles principaux
- **Patient** : Données personnelles et médicales
- **Prediction** : Résultat IA lié à un patient

````

````python
## 5. API IA Python (FestApi)
- Endpoint principal : `/predict` (POST)
- Reçoit : glucose, bmi, age, pedigree
- Retourne : cluster (0/1), catégorie de risque

### Exemple de requête
```python
import requests
data = {"Glucose": 120, "BMI": 25.3, "Age": 45, "DiabetesPedigreeFunction": 0.5}
r = requests.post('http://localhost:9000/predict', json=data)
print(r.json())
```

````

````python
## 6. Statistiques sur le code
- Fichiers Laravel : app/Http/Controllers, app/Models, resources/views
- Fichiers IA : ModelAI/api_test_model.py, .pkl
- Nombre de migrations : 2 principales (patients, predictions)
- Nombre de vues principales : 4 (liste, ajout, édition, détails)

````
